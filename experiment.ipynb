### Define Preprocessing Function


````python
# --- 1. HISTOGRAM EQUALIZATION ---
def hist_eq_scratch(img):
    hist, _ = np.histogram(img.flatten(), 256, [0,256])
    cdf = hist.cumsum()
    cdf_m = np.ma.masked_equal(cdf, 0)
    cdf_m = (cdf_m - cdf_m.min()) * 255 / (cdf_m.max() - cdf_m.min())
    cdf = np.ma.filled(cdf_m, 0).astype('uint8')
    return cdf[img]

# --- 2. GAUSSIAN BLUR ---
def gaussian_kernel(size, sigma=1.0):
    ax = np.linspace(-(size - 1) / 2., (size - 1) / 2., size)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-0.5 * (np.square(xx) + np.square(yy)) / np.square(sigma))
    return kernel / np.sum(kernel)

def gaussian_blur_scratch(img, k=3, sigma=1.0):
    kernel = gaussian_kernel(k, sigma)
    pad = k // 2
    padded = np.pad(img, pad, mode='reflect')
    windows = sliding_window_view(padded, (k, k))
    return np.sum(windows * kernel, axis=(2, 3)).astype(np.uint8)

# --- 3. PSEUDO-CLAHE (Block-wise AHE dengan Clipping) ---
def clahe_scratch(img, clip_limit=2.0, grid_size=(8, 8)):
    h, w = img.shape
    h_step, w_step = h // grid_size[0], w // grid_size[1]
    result = np.zeros_like(img)

    for i in range(grid_size[0]):
        for j in range(grid_size[1]):
            y0, y1 = i*h_step, (i+1)*h_step if i < grid_size[0]-1 else h
            x0, x1 = j*w_step, (j+1)*w_step if j < grid_size[1]-1 else w
            
            block = img[y0:y1, x0:x1]
            hist, _ = np.histogram(block.flatten(), 256, [0,256])
            
            # Clipping
            clip_val = int(clip_limit * (block.size / 256))
            excess = 0
            for k in range(256):
                if hist[k] > clip_val:
                    excess += hist[k] - clip_val
                    hist[k] = clip_val
            hist += excess // 256
            
            # Ekualisasi per blok
            cdf = hist.cumsum()
            cdf_m = np.ma.masked_equal(cdf, 0)
            cdf_m = (cdf_m - cdf_m.min()) * 255 / (cdf_m.max() - cdf_m.min())
            cdf = np.ma.filled(cdf_m, 0).astype('uint8')
            result[y0:y1, x0:x1] = cdf[block]
    return result

# --- 4. MEDIAN BLUR ---
def median_blur_scratch(img, k=3):
    pad = k // 2
    padded = np.pad(img, pad, mode='reflect')
    windows = sliding_window_view(padded, (k, k))
    return np.median(windows, axis=(2, 3)).astype(np.uint8)

# --- 5. OTSU THRESHOLDING ---
def otsu_scratch(img):
    hist, _ = np.histogram(img.flatten(), 256, [0,256])
    total = img.size
    current_max, threshold = 0, 0
    sum_total = np.dot(np.arange(256), hist)
    
    sum_B, weight_B = 0, 0
    for i in range(256):
        weight_B += hist[i]
        if weight_B == 0: continue
        weight_F = total - weight_B
        if weight_F == 0: break
        
        sum_B += i * hist[i]
        mB = sum_B / weight_B
        mF = (sum_total - sum_B) / weight_F
        
        var_between = weight_B * weight_F * (mB - mF) ** 2
        if var_between > current_max:
            current_max = var_between
            threshold = i
            
    return np.where(img >= threshold, 255, 0).astype(np.uint8)

# --- 6. MORPHOLOGY OPENING ---
def erode_scratch(img, k=3):
    pad = k // 2
    padded = np.pad(img, pad, mode='constant', constant_values=255)
    windows = sliding_window_view(padded, (k, k))
    return np.min(windows, axis=(2, 3)).astype(np.uint8)

def dilate_scratch(img, k=3):
    pad = k // 2
    padded = np.pad(img, pad, mode='constant', constant_values=0)
    windows = sliding_window_view(padded, (k, k))
    return np.max(windows, axis=(2, 3)).astype(np.uint8)

def opening_scratch(img, k=3):
    eroded = erode_scratch(img, k)
    return dilate_scratch(eroded, k)

# --- 7. BITWISE AND ---
def bitwise_and_scratch(img, mask):
    return np.where(mask > 0, img, 0).astype(np.uint8)

# --- FUNGSI RESIZE (Diizinkan pakai cv2 sesuai prompt) ---
def resize(image, target_size=(128, 128)):
    return cv.resize(image, target_size, interpolation=cv.INTER_AREA)
````

````python
def extract_and_evaluate(processed_data, labels_data, experiment_name):
    # 1. Ekstraksi GLCM
    features = []
    for img in processed_data:
        feat_row = []
        for angle in [0, 45, 90, 135]:
            g = glcm(img, angle)
            feat_row.extend([
                contrast(g), correlation(g), energy(g), 
                homogenity(g), ASM(g), entropyGlcm(g), dissimilarity(g)
            ])
        features.append(feat_row)
        
    cols = []
    for angle in [0, 45, 90, 135]:
        cols.extend([
            f'Contrast{angle}', f'Correlation{angle}', f'Energy{angle}',
            f'Homogeneity{angle}', f'ASM{angle}', f'Entropy{angle}', f'Dissimilarity{angle}'
        ])
        
    df_features = pd.DataFrame(features, columns=cols)
    
    # 2. Train-Test Split (80/20) - Konsisten
    X_train, X_test, y_train, y_test = train_test_split(
        df_features, labels_data, test_size=0.2, random_state=42
    )
    
    # 3. Normalisasi Mean-Std
    X_train_mean, X_train_std = X_train.mean(), X_train.std()
    X_train = (X_train - X_train_mean) / X_train_std
    X_test = (X_test - X_train_mean) / X_train_std
    
    # 4. Pelatihan Model
    models = {
        'KNN': KNeighborsClassifier(n_neighbors=5),
        'SVM': SVC(kernel='rbf', random_state=42),
        'Random Forest': RandomForestClassifier(n_estimators=5, random_state=42)
    }
    
    accuracies = {}
    for name, model in models.items():
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        accuracies[name] = accuracy_score(y_test, y_pred)
        
    return accuracies

def show_samples(images, title):
    fig, axes = plt.subplots(1, 6, figsize=(15, 3))
    fig.suptitle(title, fontsize=14)
    for i, ax in enumerate(axes):
        ax.imshow(images[i], cmap='gray')
        ax.axis('off')
    plt.show()
````

````python
def glcm(image, derajat):
    if derajat == 0:
        angles = [0]
    elif derajat == 45:
        angles = [np.pi / 4]
    elif derajat == 90:
        angles = [np.pi / 2]
    elif derajat == 135:
        angles = [3 * np.pi / 4]
    else:
        raise ValueError("Invalid angle. It should be one of the following: 0, 45, 90, 135.")
    
    # mengembalikan matrix glcm dari image
    glcm = graycomatrix(image, [1], angles, 256, symmetric=True, normed=True)
    return glcm
````

````python
def correlation(matriks):
	return graycoprops(matriks, 'correlation')[0, 0]
````

````python
def dissimilarity(matriks):
    return graycoprops(matriks, 'dissimilarity')[0, 0]
````

````python
def homogenity(matriks):
    return graycoprops(matriks, 'homogeneity')[0, 0]
````

````python
def contrast(matriks):
    return graycoprops(matriks, 'contrast')[0, 0]
````

````python
def ASM(matriks):
    return graycoprops(matriks, 'ASM')[0, 0]
````

````python
def energy(matriks):
    return graycoprops(matriks, 'energy')[0, 0]
````

````python
def entropyGlcm(matriks):
    return entropy(matriks.ravel())
````

````python
# Array untuk menyimpan komparasi akhir
comparison_results = []

def baseline():
    print("Menjalankan Baseline...")
    processed_imgs = []
    for img in data:
        img_resized = resize(img, (128, 128))
        # Hanya grayscale (sudah dilakukan di awal load, namun dipastikan ukurannya seragam)
        processed_imgs.append(img_resized)
        
    show_samples(processed_imgs[:6], "Sampel Citra - Baseline")
    acc = extract_and_evaluate(processed_imgs, labels, "Baseline")
    
    comparison_results.append({'Eksperimen': 'Baseline', **acc})
    print("Baseline selesai.\n")

def prepro1():
    print("Menjalankan Eksperimen 1 (HistEq + Gaussian Blur)...")
    processed_imgs = []
    for img in data:
        img_resized = resize(img, (128, 128))
        # 1. Histogram Equalization (Scratch)
        eq_img = hist_eq_scratch(img_resized)
        # 2. Gaussian Blur (Scratch)
        blur_img = gaussian_blur_scratch(eq_img, k=3, sigma=1.0)
        processed_imgs.append(blur_img)
        
    show_samples(processed_imgs[:6], "Sampel Citra - Prepro 1")
    acc = extract_and_evaluate(processed_imgs, labels, "Prepro 1")
    
    comparison_results.append({'Eksperimen': 'Prepro 1', **acc})
    print("Eksperimen 1 selesai.\n")

def prepro2():
    print("Menjalankan Eksperimen 2 (CLAHE + Median Blur)...")
    processed_imgs = []
    for img in data:
        img_resized = resize(img, (128, 128))
        # 1. CLAHE (Scratch)
        clahe_img = clahe_scratch(img_resized, clip_limit=2.0, grid_size=(8,8))
        # 2. Median Blur (Scratch)
        median_img = median_blur_scratch(clahe_img, k=3)
        processed_imgs.append(median_img)
        
    show_samples(processed_imgs[:6], "Sampel Citra - Prepro 2")
    acc = extract_and_evaluate(processed_imgs, labels, "Prepro 2")
    
    comparison_results.append({'Eksperimen': 'Prepro 2', **acc})
    print("Eksperimen 2 selesai.\n")

def prepro3():
    print("Menjalankan Eksperimen 3 (Otsu + Opening + Bitwise AND)...")
    processed_imgs = []
    for img in data:
        img_resized = resize(img, (128, 128))
        # 1. Otsu Thresholding (Scratch)
        mask = otsu_scratch(img_resized)
        # 2. Morphology Opening (Scratch)
        clean_mask = opening_scratch(mask, k=3)
        # 3. Bitwise AND (Scratch)
        segmented = bitwise_and_scratch(img_resized, clean_mask)
        processed_imgs.append(segmented)
        
    show_samples(processed_imgs[:6], "Sampel Citra - Prepro 3")
    acc = extract_and_evaluate(processed_imgs, labels, "Prepro 3")
    
    comparison_results.append({'Eksperimen': 'Prepro 3', **acc})
    print("Eksperimen 3 selesai.\n")

# --- Eksekusi Keseluruhan Pipeline ---
baseline()
prepro1()
prepro2()
prepro3()

# Menampilkan Tabel Komparasi Akhir
# df_comparison = pd.DataFrame(comparison_results)
# print("================ TABEL KOMPARASI FINAL ================")
# print(df_comparison.to_string(index=False))
# print("=======================================================")

# 1. Masukkan data hasil eksekusi pipeline Anda ke dalam DataFrame awal
df_raw = pd.DataFrame(comparison_results)

# 2. Balik posisi matriks (Transpose) agar nama eksperimen menjadi kolom di atas
df_transposed = df_raw.set_index('Eksperimen').T
df_transposed = df_transposed.reset_index()

# 3. Berikan nama kolom yang formal secara dinamis sesuai skenario yang ada
kolom_baru = {'index': 'Model'}
if 'Baseline' in df_transposed.columns:
    kolom_baru['Baseline'] = 'Akurasi Baseline (%)'
if 'Prepro 1' in df_transposed.columns:
    kolom_baru['Prepro 1'] = 'Akurasi Eksperimen 1 (%)'
if 'Prepro 2' in df_transposed.columns:
    kolom_baru['Prepro 2'] = 'Akurasi Eksperimen 2 (%)'
if 'Prepro 3' in df_transposed.columns:
    kolom_baru['Prepro 3'] = 'Akurasi Eksperimen 3 (%)'

df_transposed.rename(columns=kolom_baru, inplace=True)

# 4. Sesuaikan nama model agar formal dan presisi
model_mapping = {
    'KNN': 'K-Nearest Neighbors (KNN)',
    'SVM': 'Support Vector Machine (SVM)',
    'Random Forest': 'Random Forest'
}
df_transposed['Model'] = df_transposed['Model'].map(model_mapping)

# 5. Ubah nilai akurasi desimal ke format string persentase (2 digit di belakang koma)
for col in df_transposed.columns:
    if col != 'Model':
        df_transposed[col] = df_transposed[col].apply(lambda x: f"{x*100:.2f}" if isinstance(x, (int, float)) else x)

print("=== Tabel Perbandingan Akurasi (Baseline vs Eksperimen) ===")
# Menggunakan display() agar tabel DataFrame dirender dengan UI yang rapi di Jupyter Notebook
display(df_transposed)
````
